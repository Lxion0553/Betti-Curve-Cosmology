```python
def files_in_directory(root_dir, target):
    file_paths = []
    # loop over all subdirectories
    for root, dirs, files in os.walk(root_dir):
        for file in files: 
            if file == target: 
                
                file_path = os.path.join(root, file)  
                file_paths.append(file_path)

    return file_paths
```
